```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
from pathlib import Path

data_path = Path("../data/employee_attrition.csv")

if not data_path.exists():
    data_path = Path("backend/data/employee_attrition.csv")

df = pd.read_csv(data_path)

df.head()
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Attrition")

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
attrition_percentage = (
    df["Attrition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(attrition_percentage)
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Attrition", y="Age")

plt.title("Age Distribution by Attrition")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income vs Attrition")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="OverTime",
    hue="Attrition"
)

plt.title("OverTime vs Employee Attrition")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="JobSatisfaction",
    hue="Attrition"
)

plt.title("Job Satisfaction vs Attrition")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="YearsAtCompany"
)

plt.title("Years at Company vs Attrition")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="JobRole",
    hue="Attrition"
)

plt.title("Job Role vs Employee Attrition")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="WorkLifeBalance",
    hue="Attrition"
)

plt.title("Work-Life Balance vs Attrition")

plt.show()
```

Correlation Analysis

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])
```

```python
plt.figure(figsize=(18, 12))

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Numerical Features")

plt.show()
```

# EDA Findings

- The dataset contains employee demographic, job-related, and satisfaction-related features.
- Attrition classes are imbalanced, with more employees staying than leaving.
- Features such as OverTime, MonthlyIncome, JobSatisfaction, Age, and YearsAtCompany show meaningful relationships with Attrition.
- Some columns may not provide useful predictive information and will be evaluated during Feature Engineering and Feature Selection.
